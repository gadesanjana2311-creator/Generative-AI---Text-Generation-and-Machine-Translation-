**Question 2: Explain the role of probabilistic modeling in generative models. How do these models differ from discriminative models?**

**Answer:-**

Role of Probabilistic Modeling in Generative Models

Probabilistic modeling is the foundation of generative models. It involves using probability distributions to represent how data is generated in the real world. Generative models learn the underlying patterns, structure, and relationships in data so they can create new samples similar to the training data.

In simple terms, probabilistic modeling helps a generative model answer:

“How was this data likely generated?”

Generative models estimate the joint probability distribution of input data and labels. P(X,Y)

or sometimes only the probability distribution of the data:
P(X)

where:

X = input data

Y = labels/classes

Key Roles of Probabilistic Modeling

1. Learning Data Distribution

The model learns the probability distribution of training data, allowing it to generate realistic new examples.

Example:

A text model learns the probability of word sequences.

An image model learns pixel patterns and object structures.

2. Handling Uncertainty

Probabilistic approaches allow models to deal with

uncertainty and incomplete information effectively.

Example:

Predicting missing words in a sentence.

Generating possible future outcomes.

3. Data Generation

Once the distribution is learned, the model can sample from it to create new data.

Examples:

Generating human-like text

Creating synthetic images

Producing speech or music

4. Density Estimation

Generative models estimate how likely a data point is under the learned distribution.

This is useful in:

Anomaly detection

Image reconstruction

Data compression

5. Representation Learning

Probabilistic models often learn hidden or latent representations of data.

Examples:

Variational Autoencoders (VAEs)

Hidden Markov Models (HMMs)

**Introduction to Neural Machine Translation (NMT)**

Neural Machine Translation (NMT) is a deep learning approach used to translate text from one language to another using neural networks. Traditional NMT systems use an encoder-decoder architecture:

Encoder converts the source sentence into a context representation.
Decoder generates the translated sentence word by word.

However, early encoder-decoder models had difficulty handling long sentences because the entire sentence had to be compressed into a single fixed-length vector. This limitation led to the development of the attention mechanism.

**What is an Attention Mechanism?**

The attention mechanism allows the decoder to focus on the most relevant parts of the source sentence while generating each target word.

Instead of relying on one fixed context vector, attention dynamically selects important words from the input sequence during translation.
